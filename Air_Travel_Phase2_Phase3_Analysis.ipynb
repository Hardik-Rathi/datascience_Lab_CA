## Phase 2 – Statistical Analysis

This section covers central tendency, dispersion, skewness and kurtosis, IQR/outliers, yearly and quarterly passenger demand, correlations, and passenger growth.

```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# PHASE 2 - STATISTICAL ANALYSIS
# ============================================================

df = pd.read_csv("export_phase1_cleaned.csv")

print("Number of records:", len(df))
print("Number of columns:", len(df.columns))


# ============================================================
# SET 1 - MEAN, MEDIAN AND MODE
# ============================================================

print("\n========================================")
print("SET 1 - CENTRAL TENDENCY")
print("========================================")

columns = ["cur_passengers", "cur_fare", "cur_yield", "distance"]

for column in columns:

    values = []

    for value in df[column]:
        if pd.notna(value):
            values.append(float(value))

    n = len(values)

    total = 0
    for value in values:
        total += value

    mean = total / n

    values.sort()

    if n % 2 == 0:
        median = (values[n // 2 - 1] + values[n // 2]) / 2
    else:
        median = values[n // 2]

    frequency = {}

    for value in values:
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

    highest_frequency = 0
    mode = values[0]

    for value in frequency:
        if frequency[value] > highest_frequency:
            highest_frequency = frequency[value]
            mode = value

    print("\n", column)
    print("Mean:", mean)
    print("Median:", median)
    print("Mode:", mode)


# ============================================================
# SET 2 - RANGE, VARIANCE AND STANDARD DEVIATION
# ============================================================

print("\n========================================")
print("SET 2 - MEASURES OF DISPERSION")
print("========================================")

for column in columns:

    values = []

    for value in df[column]:
        if pd.notna(value):
            values.append(float(value))

    n = len(values)

    total = 0
    for value in values:
        total += value

    mean = total / n

    minimum = values[0]
    maximum = values[0]

    for value in values:
        if value < minimum:
            minimum = value
        if value > maximum:
            maximum = value

    range_value = maximum - minimum

    squared_difference = 0

    for value in values:
        difference = value - mean
        squared_difference += difference * difference

    variance = squared_difference / (n - 1)
    standard_deviation = variance ** 0.5

    print("\n", column)
    print("Minimum:", minimum)
    print("Maximum:", maximum)
    print("Range:", range_value)
    print("Variance:", variance)
    print("Standard Deviation:", standard_deviation)


# ============================================================
# SET 3 - SKEWNESS AND KURTOSIS
# ============================================================

print("\n========================================")
print("SET 3 - SKEWNESS AND KURTOSIS")
print("========================================")

for column in columns:

    values = []

    for value in df[column]:
        if pd.notna(value):
            values.append(float(value))

    n = len(values)

    total = 0
    for value in values:
        total += value

    mean = total / n

    squared_sum = 0
    cubed_sum = 0
    fourth_sum = 0

    for value in values:
        difference = value - mean
        squared_sum += difference ** 2
        cubed_sum += difference ** 3
        fourth_sum += difference ** 4

    variance = squared_sum / n
    standard_deviation = variance ** 0.5

    skewness = (cubed_sum / n) / (standard_deviation ** 3)
    kurtosis = (fourth_sum / n) / (standard_deviation ** 4) - 3

    print("\n", column)
    print("Skewness:", skewness)
    print("Kurtosis:", kurtosis)


# ============================================================
# SET 4 - IQR AND OUTLIERS
# ============================================================

print("\n========================================")
print("SET 4 - IQR AND OUTLIERS")
print("========================================")

values = []

for value in df["cur_passengers"]:
    if pd.notna(value):
        values.append(float(value))

values.sort()

n = len(values)

position_q1 = (n + 1) * 0.25
position_q3 = (n + 1) * 0.75

index_q1 = int(position_q1) - 1
index_q3 = int(position_q3) - 1

Q1 = values[index_q1]
Q3 = values[index_q3]

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outlier_count = 0

for value in values:
    if value < lower_limit or value > upper_limit:
        outlier_count += 1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of outliers:", outlier_count)


# ============================================================
# SET 5 - YEARLY PASSENGER DEMAND
# ============================================================

print("\n========================================")
print("SET 5 - YEARLY PASSENGER DEMAND")
print("========================================")

years = []
passenger_totals = []

for i in range(len(df)):

    year = df["Year"].iloc[i]
    passengers = df["cur_passengers"].iloc[i]

    if pd.isna(passengers):
        continue

    found = False

    for j in range(len(years)):
        if years[j] == year:
            passenger_totals[j] += passengers
            found = True
            break

    if not found:
        years.append(year)
        passenger_totals.append(passengers)

for i in range(len(years)):
    for j in range(i + 1, len(years)):
        if years[i] > years[j]:
            years[i], years[j] = years[j], years[i]
            passenger_totals[i], passenger_totals[j] = passenger_totals[j], passenger_totals[i]

for i in range(len(years)):
    print(years[i], passenger_totals[i])


# ============================================================
# SET 6 - QUARTERLY PASSENGER DEMAND
# ============================================================

print("\n========================================")
print("SET 6 - QUARTERLY PASSENGER DEMAND")
print("========================================")

quarters = []
quarter_totals = []

for i in range(len(df)):

    quarter = df["quarter"].iloc[i]
    passengers = df["cur_passengers"].iloc[i]

    if pd.isna(passengers):
        continue

    found = False

    for j in range(len(quarters)):
        if quarters[j] == quarter:
            quarter_totals[j] += passengers
            found = True
            break

    if not found:
        quarters.append(quarter)
        quarter_totals.append(passengers)

for i in range(len(quarters)):
    print(quarters[i], quarter_totals[i])


# ============================================================
# SET 7 - DISTANCE VS PASSENGERS
# ============================================================

print("\n========================================")
print("SET 7 - DISTANCE VS PASSENGERS")
print("========================================")

distance = []
passengers = []

for i in range(len(df)):

    d = df["distance"].iloc[i]
    p = df["cur_passengers"].iloc[i]

    if pd.notna(d) and pd.notna(p):
        distance.append(float(d))
        passengers.append(float(p))

n = len(distance)

sum_x = 0
sum_y = 0

for i in range(n):
    sum_x += distance[i]
    sum_y += passengers[i]

mean_x = sum_x / n
mean_y = sum_y / n

numerator = 0
sum_x_squared = 0
sum_y_squared = 0

for i in range(n):

    x_difference = distance[i] - mean_x
    y_difference = passengers[i] - mean_y

    numerator += x_difference * y_difference
    sum_x_squared += x_difference ** 2
    sum_y_squared += y_difference ** 2

correlation = numerator / ((sum_x_squared ** 0.5) * (sum_y_squared ** 0.5))

print("Correlation between distance and passengers:", correlation)


# ============================================================
# SET 8 - FARE VS PASSENGERS
# ============================================================

print("\n========================================")
print("SET 8 - FARE VS PASSENGERS")
print("========================================")

fare = []
passengers = []

for i in range(len(df)):

    f = df["cur_fare"].iloc[i]
    p = df["cur_passengers"].iloc[i]

    if pd.notna(f) and pd.notna(p):
        fare.append(float(f))
        passengers.append(float(p))

n = len(fare)

sum_x = 0
sum_y = 0

for i in range(n):
    sum_x += fare[i]
    sum_y += passengers[i]

mean_x = sum_x / n
mean_y = sum_y / n

numerator = 0
sum_x_squared = 0
sum_y_squared = 0

for i in range(n):

    x_difference = fare[i] - mean_x
    y_difference = passengers[i] - mean_y

    numerator += x_difference * y_difference
    sum_x_squared += x_difference ** 2
    sum_y_squared += y_difference ** 2

correlation = numerator / ((sum_x_squared ** 0.5) * (sum_y_squared ** 0.5))

print("Correlation between fare and passengers:", correlation)


# ============================================================
# SET 9 - DISTANCE VS FARE
# ============================================================

print("\n========================================")
print("SET 9 - DISTANCE VS FARE")
print("========================================")

distance = []
fare = []

for i in range(len(df)):

    d = df["distance"].iloc[i]
    f = df["cur_fare"].iloc[i]

    if pd.notna(d) and pd.notna(f):
        distance.append(float(d))
        fare.append(float(f))

n = len(distance)

sum_x = 0
sum_y = 0

for i in range(n):
    sum_x += distance[i]
    sum_y += fare[i]

mean_x = sum_x / n
mean_y = sum_y / n

numerator = 0
sum_x_squared = 0
sum_y_squared = 0

for i in range(n):

    x_difference = distance[i] - mean_x
    y_difference = fare[i] - mean_y

    numerator += x_difference * y_difference
    sum_x_squared += x_difference ** 2
    sum_y_squared += y_difference ** 2

correlation = numerator / ((sum_x_squared ** 0.5) * (sum_y_squared ** 0.5))

print("Correlation between distance and fare:", correlation)


# ============================================================
# SET 10 - YEAR-OVER-YEAR PASSENGER GROWTH
# ============================================================

print("\n========================================")
print("SET 10 - PASSENGER GROWTH")
print("========================================")

growth = []

for i in range(len(df)):

    current = df["cur_passengers"].iloc[i]
    previous = df["ly_passengers"].iloc[i]

    if pd.notna(current) and pd.notna(previous):

        if previous > 1000:

            percentage = ((current - previous) / previous) * 100
            growth.append(percentage)

total = 0

for value in growth:
    total += value

average_growth = total / len(growth)

growth.sort()

n = len(growth)

if n % 2 == 0:
    median_growth = (growth[n // 2 - 1] + growth[n // 2]) / 2
else:
    median_growth = growth[n // 2]

print("Average passenger growth:", average_growth)
print("Median passenger growth:", median_growth)

print("\n========================================")
print("PHASE 2 COMPLETED")
print("========================================")
```

## Phase 3 – Visualization

This section creates the histogram, boxplot, yearly passenger trend, quarterly demand chart, distance-vs-passengers scatter plot, and fare-vs-passengers scatter plot.

```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# PHASE 3 - VISUALIZATION
# ============================================================

df = pd.read_csv("export_phase1_cleaned.csv")


# ============================================================
# 1. HISTOGRAM - PASSENGER DISTRIBUTION
# ============================================================

values = []

for x in df["cur_passengers"]:
    if pd.notna(x):
        values.append(float(x))

plt.figure(figsize=(8, 5))
plt.hist(values, bins=30)
plt.xlabel("Current Passengers")
plt.ylabel("Frequency")
plt.title("Passenger Volume Distribution")
plt.show()


# ============================================================
# 2. BOXPLOT - OUTLIER ANALYSIS
# ============================================================

plt.figure(figsize=(8, 5))
plt.boxplot(values)
plt.ylabel("Current Passengers")
plt.title("Passenger Volume Outlier Analysis")
plt.show()


# ============================================================
# 3. YEARLY PASSENGER DEMAND
# ============================================================

years = []
passengers = []

for i in range(len(df)):

    if pd.notna(df["Year"][i]) and pd.notna(df["cur_passengers"][i]):

        year = int(df["Year"][i])
        value = float(df["cur_passengers"][i])

        if year in years:

            index = years.index(year)
            passengers[index] += value

        else:

            years.append(year)
            passengers.append(value)


for i in range(len(years)):

    for j in range(i + 1, len(years)):

        if years[i] > years[j]:

            years[i], years[j] = years[j], years[i]
            passengers[i], passengers[j] = passengers[j], passengers[i]


plt.figure(figsize=(9, 5))
plt.plot(years, passengers, marker="o")
plt.xlabel("Year")
plt.ylabel("Total Passengers")
plt.title("Passenger Demand Over Time")
plt.grid()
plt.show()


# ============================================================
# 4. QUARTERLY PASSENGER DEMAND
# ============================================================

quarters = [1, 2, 3, 4]
quarter_passengers = [0, 0, 0, 0]

for i in range(len(df)):

    if pd.notna(df["quarter"][i]) and pd.notna(df["cur_passengers"][i]):

        q = int(df["quarter"][i])
        quarter_passengers[q - 1] += float(df["cur_passengers"][i])


plt.figure(figsize=(8, 5))
plt.bar(quarters, quarter_passengers)
plt.xlabel("Quarter")
plt.ylabel("Total Passengers")
plt.title("Passenger Demand by Quarter")
plt.show()


# ============================================================
# 5. DISTANCE VS PASSENGERS
# ============================================================

distance = []
passenger = []

for i in range(len(df)):

    if pd.notna(df["distance"][i]) and pd.notna(df["cur_passengers"][i]):

        distance.append(float(df["distance"][i]))
        passenger.append(float(df["cur_passengers"][i]))


plt.figure(figsize=(8, 5))
plt.scatter(distance, passenger, s=10)
plt.xlabel("Distance")
plt.ylabel("Current Passengers")
plt.title("Distance vs Passenger Demand")
plt.show()


# ============================================================
# 6. FARE VS PASSENGERS
# ============================================================

fare = []
passenger = []

for i in range(len(df)):

    if pd.notna(df["cur_fare"][i]) and pd.notna(df["cur_passengers"][i]):

        fare.append(float(df["cur_fare"][i]))
        passenger.append(float(df["cur_passengers"][i]))


plt.figure(figsize=(8, 5))
plt.scatter(fare, passenger, s=10)
plt.xlabel("Current Fare")
plt.ylabel("Current Passengers")
plt.title("Fare vs Passenger Demand")
plt.show()


print("Phase 3 completed successfully.")
```

## Key interpretation checklist

- **Histogram:** inspect the distribution and skewness of passenger volume.
- **Boxplot:** inspect spread and high-volume outliers.
- **Yearly trend:** identify major increases, decreases, and recovery periods.
- **Quarterly demand:** compare Q1, Q2, Q3 and Q4.
- **Distance vs passengers:** assess whether route distance is associated with passenger demand.
- **Fare vs passengers:** assess whether fare is associated with passenger demand.
